```python
# ============================================================
# FINACCESS 2024 — FINANCIAL VARIABLE SHORTLIST
# ============================================================

keywords = [
    "loan",
    "credit",
    "borrow",
    "borrowing",
    "save",
    "savings",
    "saving",
    "income",
    "expense",
    "expenditure",
    "debt",
    "insurance",
    "financial",
    "bank",
    "mobile money",
    "mobile",
    "money",
    "shock",
    "vulnerability",
    "access",
    "account",
    "investment",
    "invest",
    "pension",
    "wealth",
    "payment"
]

matches = []

# Search both variable names and official variable labels
for var, label in zip(meta.column_names, meta.column_labels):

    var_text = str(var)
    label_text = str(label)

    combined_text = f"{var_text} {label_text}".lower()

    if any(keyword in combined_text for keyword in keywords):
        matches.append((var_text, label_text))


# ============================================================
# REMOVE DUPLICATES
# ============================================================

unique_matches = []
seen = set()

for var, label in matches:

    if var not in seen:
        unique_matches.append((var, label))
        seen.add(var)


# ============================================================
# DISPLAY RESULTS
# ============================================================

print("=" * 100)
print("FINACCESS 2024 — POTENTIALLY RELEVANT FINANCIAL VARIABLES")
print("=" * 100)

print(f"\nTotal matching variables: {len(unique_matches)}\n")

for number, (var, label) in enumerate(unique_matches, start=1):

    print(f"{number:3}. {var:25} | {label}")

print("\n" + "=" * 100)
print("Inspection complete.")
print("=" * 100)
```

```python
keywords = [
    "income",
    "employment",
    "employed",
    "occupation",
    "business",
    "household",
    "depend",
    "financial literacy",
    "literacy",
    "debt",
    "credit",
    "loan",
    "saving",
    "shock",
    "financial access",
    "access"
]

for keyword in keywords:
    print(f"\n{'='*70}")
    print(f"KEYWORD: {keyword.upper()}")

    matches = []

    for var, label in meta.column_names_to_labels.items():
        if label and keyword.lower() in str(label).lower():
            matches.append((var, label))

    for var, label in matches:
        print(f"{var:35} | {label}")
```

```python
keywords = {
    "Income": ["income", "earnings"],
    "Employment": ["employment", "employed", "occupation", "job"],
    "Household": ["household size", "household members", "depend"],
    "Business": ["business owner", "business activity", "enterprise"],
    "Debt": ["debt", "outstanding"],
    "Financial literacy": ["financial literacy"],
    "Financial access": ["financial access"],
    "Shock": ["shock"]
}

for category, terms in keywords.items():
    matches = []

    for var, label in meta.column_names_to_labels.items():
        if label:
            label_lower = str(label).lower()
            if any(term.lower() in label_lower for term in terms):
                matches.append((var, label))

    print(f"\n{category}: {len(matches)} matches")
    for var, label in matches[:10]:
        print(f"  {var} → {label}")
```

```python
candidate_vars = [
    "Age",
    "Sex",
    "Education",
    "NHM",
    "B3A__2",
    "B3A__4",
    "Savings_usage",
    "Savings_formal",
    "Savings_informal",
    "Loan_usage",
    "Loan_formal",
    "Loan_informal",
    "Digital_credit_usage_fnl",
    "mobile_money_access_2",
    "All_Insurance_excluding_NHIF",
    "All_Insurance_including_NHIF",
    "Pension_usage",
    "Pension_Usage_without_NSSF",
    "fl_score",
    "experienced_shock"
]

results = []

for var in candidate_vars:
    grouped = df.groupby(var)["defaulted"].agg(["mean", "count"])

    results.append({
        "variable": var,
        "groups": df[var].nunique(),
        "min_default_rate": grouped["mean"].min(),
        "max_default_rate": grouped["mean"].max()
    })

pd.DataFrame(results).sort_values("max_default_rate", ascending=False)
```
